## Folder structure

```python
import os

def save_directory_structure(root_dir, output_file='folder_structure.txt'):
    """
    Saves the complete folder and file structure of root_dir to a text file.
    
    Args:
        root_dir (str): The path to the folder you want to map (e.g., r'Z:\Guoming_Gao_turbo').
        output_file (str): The name of the output text file.
    """
    try:
        with open(output_file, 'w', encoding='utf-8') as file:
            # Write a header
            file.write(f"Directory structure of: {root_dir}\n")
            file.write("=" * 60 + "\n\n")
            
            # Walk through all directories, subdirectories, and files
            for dirpath, dirnames, filenames in os.walk(root_dir):
                # Calculate the depth for visual indentation
                level = dirpath.replace(root_dir, '').count(os.sep)
                indent = '    ' * level  # 4 spaces per level
                
                # Write the current directory name
                file.write(f"{indent}{os.path.basename(dirpath)}/\n")
                
                # Write all file names in the current directory
                for f_name in filenames:
                    file.write(f"{indent}    ├── {f_name}\n")
    
        print(f"✅ Success! Directory structure saved to: {os.path.abspath(output_file)}")
    
    except PermissionError as e:
        print(f"⚠️  Permission denied for a folder: {e}")
    except FileNotFoundError:
        print(f"❌ Error: The path '{root_dir}' does not exist.")
    except Exception as e:
        print(f"❌ An unexpected error occurred: {e}")

# --- Configuration: SET YOUR PATH HERE ---
# Replace this with the path to your folder on the server
TARGET_FOLDER = r"Z:\Guoming_Gao_turbo"
OUTPUT_FILENAME = "Guoming_Gao_turbo_structure.txt"

# Run the function
if __name__ == "__main__":
    save_directory_structure(TARGET_FOLDER, OUTPUT_FILENAME)
```
